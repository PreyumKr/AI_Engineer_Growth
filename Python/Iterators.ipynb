```python
# Iterators

# An iterator is a object that iterates over a sequence of values without having to store the entire sequence in memory.

# Every iterator is also an iterable, but not every iterable is an iterator.
```

```python
L = [1, 2, 3, 4, 5]
dir(L)
# __iter__ says its iterable
```

```python
iterator = iter(L)
dir(iterator)

# __iter__ and __next__ says its an iterator
```

```python
print(next(iterator))  # 1
print(next(iterator))  # 2
print(next(iterator))  # 3
print(next(iterator))  # 4
print(next(iterator))  # 5
# print(next(iterator))  # Error
```

```python
# Our own for loop implementation

def my_own_loop(iterable):
    iterator = iter(iterable)
    while True:
        try:
            item = next(iterator)
            print(item)
        except StopIteration:
            break

A = [1, 2, 3, 4, 5]
my_own_loop(A)
```

```python
# Creating our own iteratable class

class MyIterable:
    def __init__(self, start, end):
        self.start = start
        self.end = end

    def __iter__(self):
        return MyIterator(self.start, self.end)

class MyIterator:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self

    def __next__(self):
        # like range, we want to stop when current is equal to end
        if self.current >= self.end:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1


obj = MyIterable(0, 5)
for i in obj:
    print(i)
    
```

```python
# Simplifying the above code using generator function

def my_generator(start, end):
    current = start
    while current < end:
        yield current
        current += 1

for i in my_generator(0, 5):
    print(i)

# Difference between yield and return
# Every time we call a function, it starts from the beginning. But when we use yield, it remembers where it left off and continues from there.
```

```python
# Another example of generator function

def gen_demo():
    # yield 1
    # yield 2
    # yield 3
    yield "First Statement"
    yield "Second Statement"
    yield "Third Statement"

gen = gen_demo()

for i in gen:
    print(i)
```
