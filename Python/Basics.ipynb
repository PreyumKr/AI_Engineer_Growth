```python
# Use of Flush in Python
import time

# WITHOUT FLUSH (May pause for 3 seconds, then show everything at once)
for i in range(3, 0, -1):
    print(i, end=" ")
    time.sleep(1)

print("\n---")

# WITH FLUSH (Prints 3... then 2... then 1... exactly second by second)
for i in range(3, 0, -1):
    print(i, end=" ", flush=True)
    time.sleep(1)
```
