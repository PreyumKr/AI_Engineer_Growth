```python
# Use of Flush in Python
import time

# WITHOUT FLUSH (May pause for 3 seconds, then show everything at once)
for i in range(3, 0, -1):
    print(i, end=" ")
    time.sleep(1)

print("\n---")

# WITH FLUSH (Prints 3... then 2... then 1... exactly second by second)
for i in range(3, 0, -1):
    print(i, end=" ", flush=True)
    time.sleep(1)
```

```python
# Type conversion for supported types

# Implicit
print(4+5.5)
print(4.5+ 6+8j)

# Explicit
print(int(8.9))
print(int("787"))

# Type conversion does not change the original variable, it returns a new variable of the converted type.
x = 5
y = float(x)
print(x)
print(y)
```

```python
# is operator
# It checks if the two variables point to the same object in memory. It does not check for equality of values.

x = 5
y = 5
print(x is y) # True, because both point to the same integer object in memory

print(x == y) # True, because their values are equal

x = [1, 2, 3]
y = [1, 2, 3]
print(x is y) # False, because they are different objects in memory

print(x == y) # True, because their values are equal

# The negation of the is operator is not, which checks if two variables do not point to the same object in memory.

a = 5
b = 10
print(a is not b) # True, because they are different objects in memory
```

```python
# break, continue and pass statements

for i in range(1, 11):
    if i == 5:
        break # Exit the loop when i is 5
    print(i, end=" ")

print("\n---")

for i in range(1, 11):
    if i == 5:
        continue # Skip the rest of the loop when i is 5
    print(i, end=" ")

print("\n---")

for i in range(1, 11):
    if i == 5:
        pass # Do nothing when i is 5
    print(i, end=" ")
```

```python
# Lists can store heterogeneous data types

L = [1,2,"Hello", 3.14, True, [1, 2, 3], (4, 5), {6, 7}, {'key': 'value'}, 3+8j]
L
```

```python
# Methods in a list

L.append(100) # Adds 100 to the end of the list
print(L)

L.extend([200, 300]) # Adds multiple elements to the end of the list
print(L)

L.append("hello") # Adds "hello" to the end of the list
print(L)

L.extend("Hello") # Adds each character of the string as a separate element to the end of the list
print(L)

L.insert(2, "Inserted") # Inserts "Inserted" at index 2
print(L)
```

```python
# delete an element from a list

L.remove("Inserted") # Removes the first occurrence of "Inserted" from the list
print(L)

del L[2] # Deletes the element at index 2
print(L)

L.remove("hello") # Removes the first occurrence of "hello" from the list
print(L)

L.pop() # Removes and returns the last element of the list
print(L)

L.clear() # Removes all elements from the list
print(L)
```

```python
# dictionaries

# keys: immutable, values: mutable
# mutable: list, set, dict
# immutable: strings, tuples, int, float, boolean, complex

d = {}
print(type(d)) # <class 'dict'>
print(d) # {}

d1 = {"name": "Preyum", "age": 25, "city": "Goa"}
print(d1) 
```

```python
d2 = {[1, 2]: "value"} # This will raise a TypeError because lists are mutable and cannot be used as dictionary keys
print(d2)
```

```python
d2 = {(1, 2): "value"} # This will work because tuples are immutable and can be used as dictionary keys
print(d2)
```

```python
d3 = {"name": "Preyum", "name": "Kumar"} # name key is repeated, so the last value will overwrite the previous value
print(d3)
```

```python
# accessing items in a dictionary

d = {"name": "Preyum", "age": 25, "city": "Goa"}

print(d["name"]) # Accessing value using key
print(d.get("age")) # Accessing value using get() method
```

```python
# editing a dictionary

d['name'] = "Kumar" # Changing the value of the key 'name'
print(d)

# add a new key-value pair to the dictionary
d['country'] = "India" # Adding a new key-value pair to the dictionary
print(d)
```

```python
# delete a key-value pair from the dictionary
del d['age'] # Deleting the key-value pair with key 'age'
print(d)
```

```python
# looping in dictionaries

for key in d:
    print(key, d[key]) # Prints the key and its corresponding value

for value in d.values():
    print(value) # Prints the values in the dictionary

for key, value in d.items():
    print(key, value) # Prints the keys and their corresponding values
```
