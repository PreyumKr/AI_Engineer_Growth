## Threading in Python

```python
# Sequential access 

from time import time, sleep

def something(id):
    print(f"Task {id} started")
    sleep(1)
    print(f"Task {id} completed")

start = time()

for i in range(10):
    something(i)

end = time()

print(f"Main thread ended. Total time taken: {end - start} seconds")
```

```python
# Using threading to run tasks concurrently

import threading

threads = [threading.Thread(target=something, args=[i]) for i in range(10)]

start = time()

for thread in threads:
    thread.start()

for thread in threads:
    thread.join()

end = time()

print(f"Main thread ended. Total time taken: {end - start} seconds")
```

```python
# Race condition example

import threading
import time

counter = 0

def increment():
    global counter
    for _ in range(10000):
        temp = counter
        time.sleep(0)  # Forces a thread switch right here
        counter = temp + 1

def decrement():
    global counter
    for _ in range(10000):
        temp = counter
        time.sleep(0)  # Forces a thread switch right here
        counter = temp - 1

thread1 = threading.Thread(target=increment)
thread2 = threading.Thread(target=decrement)

thread1.start()
thread2.start()
thread1.join()
thread2.join()

print(f"Final counter value: {counter}")
```

```python
import threading
import time
from threading import Lock

counter = 0
lock = Lock()

def increment():
    global counter
    for _ in range(10000):
        with lock:
            temp = counter
            time.sleep(0)  # Forces a thread switch right here
            counter = temp + 1

def decrement():
    global counter
    for _ in range(10000):
        with lock:
            temp = counter
            time.sleep(0)  # Forces a thread switch right here
            counter = temp - 1

thread1 = threading.Thread(target=increment)
thread2 = threading.Thread(target=decrement)

thread1.start()
thread2.start()
thread1.join()
thread2.join()

print(f"Final counter value: {counter}")
```
