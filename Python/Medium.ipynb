```python
# Format methods
name = "Alice"
age = 30
# Using f-strings (Python 3.6+)
print(f"My name is {name} and I am {age} years old.")
print("My name is {} and I am {} years old.".format(name, age)) # Using str.format()
print("My name is {0} and I am {1} years old.".format(name, age)) # Using str.format() with positional arguments
print("My name is {name} and I am {age} years old.".format(name=name, age=age)) # Using str.format() with keyword arguments
```

```python
# isalnum, isalpha, isdigit, islower, isupper, isspace, isidentifier methods
print("hello".isalnum()) # True, because all characters are alphanumeric
print("hello123".isalnum()) # True, because all characters are alphanumeric
print("hello!".isalnum()) # False, because '!' is not alphanumeric
print("hello".isalpha()) # True, because all characters are alphabetic
print("hello123".isalpha()) # False, because '1', '2', and '3' are not alphabetic
print("123".isdigit()) # True, because all characters are digits
print("hello".isdigit()) # False, because 'h', 'e', 'l', 'l', and 'o' are not digits
print("hello".islower()) # True, because all characters are lowercase
print("HELLO".isupper()) # True, because all characters are uppercase
print("   ".isspace()) # True, because all characters are whitespace
print("hello world".isspace()) # False, because not all characters are whitespace
print("hello".isidentifier()) # True, because 'hello' is a valid identifier
print("123hello".isidentifier()) # False, because identifiers cannot start with a digit
```

```python
# Sets

# They dont have duplicates and no indexing or slicing
# Sets dont allow mutable elements like lists or dictionaries as elements, but they can have immutable elements like tuples.
# Since sets are mutable themselves, we cant have 2d sets or sets of sets, but we can have sets of tuples.

s1 = {} # does not create an empty set, it creates an empty dictionary
print(type(s1)) # <class 'dict'>

s2 = set() # creates an empty set
print(type(s2)) # <class 'set'>
```

```python
s3 = {"hello", 1, 2,3,3,2} # creates a set with unique elements
print(s3) # {'hello', 1, 2, 3}
```

```python
s4 = {[1, 2], 3} # This will raise a TypeError because lists are mutable and cannot be added to a set
```

```python
s5 = {(1, 2), 3, "Hello"}
print(s5) # order of elements in a set is not guaranteed, so the output may vary
```

```python
# Add items in set 

s5.add(6) # adds 6 to the set
print(s5) # order of elements in a set is not guaranteed, so the output may vary
```

```python
# remove items from set

s5.remove(3) # removes 3 from the set, raises KeyError if 3 is not in the set
print(s5) 

s5.pop() # removes arbitrary element from the set, raises KeyError if the set is empty 
print(s5)
```

```python
# looping in sets

s = {1, "apple", 3.14, (1, 2), True}
print(s) # true is not shown as 1 and true are treated as the same value in a set
for item in s:
    print(item)
```

```python
# methods only in sets

s1 = {1, 2, 3}
s2 = {3, 4, 5}
s3 = {4, 5, 6}

print(s1.union(s2)) # {1, 2, 3, 4, 5}
print(s1.intersection(s2)) # {3}
print(s1.difference(s2)) # {1, 2}
print(s2.difference(s1)) # {4, 5}
print(s1.symmetric_difference(s2)) # {1, 2, 4, 5}
print(s1.isdisjoint(s3)) # True, because s1 and s3 have no elements in common
print(s1.isdisjoint(s2)) # False, because s1 and s2 have the element 3 in common
print(s1.issubset(s2)) # False, because s1 is not a subset of s2
print(s1.issuperset(s2)) # False, because s1 is not a superset of s2
s1.clear() # removes all elements from the set
print(s1) # set()
```

```python
# Issues with copying mutable objects

L = [1, 2, 3]
L1 = L # This creates a new reference to the same list object in memory, so both L and L1 point to the same list object. This is called shallow copying.

L1.append(4) # This modifies the list object that both L and L1 point to, so both L and L1 will reflect this change.
print(L) # [1, 2, 3, 4]
print(L1) # [1, 2, 3, 4]
```

```python
# Instead, we can use the copy module to create a deep copy of the list object, so that L and L1 point to different list objects in memory.

X = [1, 2, 3]
import copy
Y = copy.deepcopy(X) # This creates a new list object in memory, so both X and Y point to different list objects in memory. This is called deep copying.
Y.append(4) # This modifies the list object that Y points to, so only Y will reflect this change.
print(X) # [1, 2, 3]
print(Y) # [1, 2, 3, 4]
```

```python
# Simple way to clone a list is to use the list() constructor or the slicing operator [:]

Z = [1, 2, 3]
Z1 = list(Z) # This creates a new list object in memory, so both Z

Z1.append(4) # This modifies the list object that Z1 points to, so only Z1 will reflect this change.

print(Z) # [1, 2, 3]
print(Z1) # [1, 2, 3, 4]

Q = [1, 2, 3]
Q1 = Q[:] # This creates a new list object in memory, so both Q and Q1 point to different list objects in memory. This is called deep copying.
Q1.append(4) # This modifies the list object that Q1 points to, so only Q1 will reflect this change.

print(Q) # [1, 2, 3]
print(Q1) # [1, 2, 3, 4]
```

```python
# Creating function with variable number of arguments

def prod(*args):
    result = 1
    for arg in args:
        result *= arg
    return result

print(prod(1, 2, 3)) # 6
print(prod(1, 2, 3, 4)) # 24
print(prod(1, 2, 3, 4, 5)) # 120

# Argument vs Parameter
# Parameters are the variables that are defined in the function definition, while arguments are the values that are passed to the function when it is called.
```

```python
# Palindrome finding recursion

def is_palindrome(s):
    # base case:
    if len(s) <= 1:
        return True
    # recursive case:
    if s[0] != s[-1]:
        return False
    return is_palindrome(s[1:-1])

print(is_palindrome("racecar")) # True
print(is_palindrome("hello")) # False
print(is_palindrome("abba")) # True
```
