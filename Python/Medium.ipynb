```python
# literals in python

# Numric literals
a = 100 # Decimal
b = 0b100 # Binary
c = 0o100 # Octal
d = 0x100 # Hexadecimal

print(a, b, c, d)
```

```python
# String literals
s1 = 'Hello\nword\n' # Normal string
s2 = u"\U0001F600" # Unicode string
s3 = r"Hello\nWorld" # Raw string
print(s1, s2, s3)
```

```python
# Boolean literals
b1 = True
b2 = False
b3 = True + 3 # True is treated as 1
b4 = False + 3 # False is treated as 0
print(b1, b2, b3, b4)
```

```python
# Special literals
n1 = None # Variable declaration
```

```python
# Ascii values of characters
print(ord('A')) # Returns the ASCII value of 'A'
print(chr(65)) # Returns the character corresponding to ASCII value 65
```

```python
# Indexing and slicing
s = "Hello, World!"
print('s[0]:', s[0]) # Returns 'H'
print('s[-1]:', s[-1]) # Returns '!'
print('s[0:5]:', s[0:5]) # Returns 'Hello' 
print('s[-1:-6:-1]:', s[-1:-6:-1]) # Cant make step in opposite direction if slicing is in different direction, it will return empty string
print('s[-1:-6:1]:', s[-1:-6:1]) 
print('s[1:6:1]:', s[1:6:1]) 
print('s[1:6:-1]:', s[1:6:-1]) 
```
